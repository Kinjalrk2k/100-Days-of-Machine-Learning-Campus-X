```python
from dotenv import load_dotenv
import os
import requests
import pandas as pd
```

```python
load_dotenv()
```

## Trying out API

```python
url = "https://api.themoviedb.org/3/movie/top_rated?language=en-US&page=1"

headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {os.getenv("TMDB_API_READ_ACCESS_TOKEN")}"
}

response = requests.get(url, headers=headers)
results = response.json()["results"]
print(results)
```

```python
df = pd.DataFrame(results)

df = df.set_index("id")
df.info()
```

```python
df["poster_path"] = df["poster_path"].map(lambda x: "https://image.tmdb.org/t/p/original"+x)
df["backdrop_path"] = df["backdrop_path"].map(lambda x: "https://image.tmdb.org/t/p/original"+x)
df.head()
```

## Creating entire dataset

```python
df = pd.DataFrame()
```

```python
for i in range(1, 497):
    url = f"https://api.themoviedb.org/3/movie/top_rated?language=en-US&page={i}"

    headers = {
        "accept": "application/json",
        "Authorization": f"Bearer {os.getenv("TMDB_API_READ_ACCESS_TOKEN")}"
    }

    response = requests.get(url, headers=headers)
    results = response.json()["results"]

    temp_df = pd.DataFrame(results)
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
# df = df.set_index("id")
df["poster_path"] = df["poster_path"].map(lambda x: "https://image.tmdb.org/t/p/original"+x if x else None)
df["backdrop_path"] = df["backdrop_path"].map(lambda x: "https://image.tmdb.org/t/p/original"+x if x else None)
```
